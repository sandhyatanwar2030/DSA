```python
# Write a program to find the sum of all no between 1-20 using for loop
sum=0
for i in range(1,21):
  sum=sum+i
print(sum)
```

```python
# Write a program to find the sum of all even no between 1-20 using for loop

sum=0
for i in range(1,21):
  if(i%2==0):
    sum=sum+i
print(sum)
```

```python
#WAP to count the no. of vowels in a given string

count=0
Str = str(input("Enter a String"))
vowels = ['a','e','i','o','u']
for i in Str:
  if i.lower() in vowels:
    count+=1
print(f"No. of Vowels in a {Str} is {count} ")
```

```python
# WAP to print the fibonacci series upto n terms

n=int(input("Enter the nth term"))

a=0
b=1

for i in range(n):
  print(a,end=",")
  c=a+b
  a=b
  b=c
```

```python
# WAP to check if the given number is prime or not

num = int(input("Enter a number: "))

if num < 2:
    print(num, "is not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
      print(num, "is a prime number")
```
