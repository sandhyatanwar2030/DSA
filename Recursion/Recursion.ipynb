 FACTORIAL using recursion function

```python
#find FACTORIAL using recursion function

def factorial(n):
  #Base Condition
  if(n==0):
    return 1
  # Recursive call
  else:
    return n*factorial(n-1)
```

```python
n=int(input("Enter the number u want to find factorial of"))
if n<0:
  print("Enter a positive value")
else:
  print(f"Factorial of {n} is {factorial(n)}")
```

find FIBONACCI series

```python
# find FIBONACCI series

def fibo(n):
  if n==0:
    return 0
  elif n==1:
    return 1
  else:
    return fibo(n-1)+fibo(n-2)
```

```python
n=int(input("Enter the number upto which u want fibonacci series"))
for i in range(n):
  print(fibo(i),end=" ")
```

Sum of array using Recursive method ...


```python
def sum_of_array(arr,n):
  #base condition
  if n==0:
    return 0
  #recursive condition
  else:
    return arr[n-1] +  sum_of_array(arr,n-1)

arr=[1,2,3,4]
print(sum_of_array(arr,len(arr)))
```

PRINT 1 To n INTEGERS

```python
#using iterative approach

n=int (input("Enter the value of n upto which u want to print: "))

for i in range(1,n+1):
  print(i ,end="\n")

```

```python
#using Recursive approach.....
def print_numbers(n):
    if n == 0:
        return

    print_numbers(n - 1)
    print(n)

n = int(input("Enter value of n: "))
print_numbers(n)
```

```python
#using iterative approach

n=int (input("Enter the value of n upto which u want to print: "))

for i in range(n,0,-1):
  print(i ,end="\n")
```

Climbimg Stairs using Recursion

```python
def Stairs(n):
  #base condition
  if n==0:
    return 1
  elif n==1:
    return 1
  elif n<0:
    return 0
  #Recursive call
  return Stairs(n-1)+Stairs(n-2)

print(Stairs(6))

#Time Omplexity = O(2^n)
#this should be done with Dynamic programming rather than using recursion for better time complexity
```

Binary Search using recursion...


```python
def Binary_Search(arr,low,high,target):
  mid=(high+low)//2
#base condition
  if low > high:
    return -1
#recursive call
  elif(arr[mid]==target):
    return mid

  elif(arr[mid]>target):
    return Binary_Search(arr,low,mid-1,target)
  else:
    return Binary_Search(arr,mid+1,high,target)


arr=[1,2,3,4,5,6,7,8]
Binary_Search(arr,0,len(arr),5)
```

Subset of String


```python
def Subset(s,current="",index=0):
  #base Condition
  if index==len(s):
    print(current)
    return
  #Recursive call
  Subset(s,current+s[index],index+1)
  Subset(s,current, index + 1)
```
